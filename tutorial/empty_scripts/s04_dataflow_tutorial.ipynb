## 1. Toy GRAPH files (what s03 writes)

s03 writes one `<burst>_GRAPH.parquet` per burst into `data/<set>/<location>/<orbit>/`. Every row is one EGMS point that
belongs to a pool. A pool = all rows with the same `seed_pid`; exactly one row of it has `seed == 1` (the seed).

The cell below builds the same structure in a temporary folder: 2 train regions and 1 dev region, a few pools each,
with a smooth fake velocity field and random fake GSE vectors. This is toy-data code, not s04 — you don't have to go
through it line by line.

```python
ROOT = Path(tempfile.mkdtemp()) / "data"
rng = np.random.default_rng(0)

def toy_velocity(lat, lon):
    # a smooth fake deformation field in mm/yr: a subsidence bowl plus a gentle trend
    return -15 * np.exp(-((lat - 48.10) ** 2 + (lon - 11.50) ** 2) / 0.0004) + 20 * (lon - 11.5)

def quantise_gse(x):
    # the inverse of the de-quantisation in s04: unit vectors -> int8, like the real GSE columns
    return np.clip(np.round(np.sign(x) * np.sqrt(np.abs(x)) * 127.5), -127, 127).astype(np.int8)

def make_graph_file(path, n_pools, centre, prefix):
    rows = []
    for p in range(n_pools):
        n = int(rng.integers(5, 27))                      # pool size incl. the seed (some below min_graph_size = 7)
        seed_lat = centre[0] + rng.normal(0, 0.01); seed_lon = centre[1] + rng.normal(0, 0.01)
        lat = seed_lat + np.r_[0, rng.normal(0, 0.005, n - 1)]   # first point = the seed, the others within ~1 km
        lon = seed_lon + np.r_[0, rng.normal(0, 0.007, n - 1)]
        gse = rng.normal(size=(n, 64)); gse /= np.linalg.norm(gse, axis=1, keepdims=True)
        seed_pid = f"{prefix}_{p}_0"
        for j in range(n):
            row = {"pid": f"{prefix}_{p}_{j}", "seed_pid": seed_pid, "seed": np.int8(j == 0),
                   "latitude": lat[j], "longitude": lon[j], "easting": 0.0, "northing": 0.0,
                   "incidence_angle": 38 + rng.normal(0, 3),
                   "mean_velocity": toy_velocity(lat[j], lon[j]) + rng.normal(0, 1)}
            row.update({f"gse_{i}": v for i, v in enumerate(quantise_gse(gse[j]))})
            rows.append(row)
    df = pl.DataFrame(rows).with_columns(pl.col("seed").cast(pl.Int8), pl.col([f"gse_{i}" for i in range(64)]).cast(pl.Int8))
    path.parent.mkdir(parents=True, exist_ok=True)
    df.write_parquet(path)
    return str(path)

train_files = [make_graph_file(ROOT / "train" / "RegionA" / "ascending" / "burstA1_GRAPH.parquet", 12, (48.10, 11.50), "A1"),
               make_graph_file(ROOT / "train" / "RegionA" / "descending" / "burstA2_GRAPH.parquet", 10, (48.10, 11.50), "A2"),
               make_graph_file(ROOT / "train" / "RegionB" / "ascending" / "burstB1_GRAPH.parquet", 12, (48.12, 11.47), "B1")]
dev_files = [make_graph_file(ROOT / "dev" / "RegionC" / "ascending" / "burstC1_GRAPH.parquet", 8, (48.11, 11.49), "C1")]
print("train files:", *train_files, sep="\n  ")
print("dev files:", *dev_files, sep="\n  ")
```

## 2. The pool index and the sample

### 2.1 `build_pool_index` — list the pools without loading their points

Instead of loading all points, s04 first only lists the pools: one row per pool, (`seed_pid`, `graph_file`). Every pool
has exactly one seed row, so the seed rows of a file are exactly its pools.

```python
def build_pool_index(graph_files):
    tables = [pl.scan_parquet(path).filter(pl.col("seed") == 1).select("seed_pid")
              .with_columns(pl.lit(str(path)).alias("graph_file")).collect() for path in graph_files]
    return pl.concat(tables)
```

**Line by line.** The whole function is one list comprehension with a polars chain inside. We take the chain apart for one file, then do the loop and the concat.

```python
# pl.scan_parquet(path): does NOT read the file. It returns a LazyFrame = a plan of what to read later.
# Nothing is in memory until .collect() runs at the end of the chain, so polars can optimise the whole chain at once.
path = train_files[0]
lazy = pl.scan_parquet(path)
show("lazy", lazy)
print("the file has", len(lazy.collect_schema().names()), "columns, the first ones:", lazy.collect_schema().names()[:8])
```

```python
# .filter(pl.col("seed") == 1): keep only the rows where the seed flag is 1 -> one row per pool.
# pl.col("seed") == 1 is an expression (a description of a computation), not a result yet.
expr = pl.col("seed") == 1
show("expr", expr)
lazy_seeds = lazy.filter(expr)
print("plan (read bottom-up):\n", lazy_seeds.explain())
print("\nwhat the filter keeps (collected only for the print):")
print(lazy_seeds.select("pid", "seed_pid", "seed").collect().head())
```

```python
# .select("seed_pid"): of the seed rows only the pool id is needed
lazy_ids = lazy_seeds.select("seed_pid")
print(lazy_ids.collect().head())
```

```python
# .with_columns(pl.lit(str(path)).alias("graph_file")): adds a column holding the same value in every row.
# pl.lit = a literal (constant) value, .alias = the column name. So every pool knows which file it comes from,
# which load_pools needs later to know where to read the pool's points.
lazy_index = lazy_ids.with_columns(pl.lit(str(path)).alias("graph_file"))
print(lazy_index.collect().head())
```

```python
# .collect(): now the plan is executed and the file is actually read -> a (small) DataFrame
table = lazy_index.collect()
show("table", table)
```

```python
# the list comprehension does the same for every file of the list -> one small DataFrame per file
tables = [pl.scan_parquet(path).filter(pl.col("seed") == 1).select("seed_pid")
          .with_columns(pl.lit(str(path)).alias("graph_file")).collect() for path in train_files]
print("type:", type(tables).__name__, "| one table per file:", len(tables), "| rows per table:", [t.height for t in tables])
```

```python
# pl.concat(tables): stacks the tables on top of each other -> one index over all train files
pool_index = pl.concat(tables)
show("pool_index", pool_index)
print("check, same as the function:", pool_index.equals(build_pool_index(train_files)))
train_index_all = pool_index
```

### 2.2 `sample_pools` — keep `--n-graphs` pools at random

Draws `n_graphs` rows of the pool index at random over all files at once, so every region keeps its share of pools on
average. With `n_graphs = None` all pools are kept.

```python
def sample_pools(pool_index, n_graphs, seed):
    if n_graphs is None or n_graphs >= pool_index.height:
        return pool_index
    rng = np.random.default_rng(seed)
    positions = rng.choice(pool_index.height, size=n_graphs, replace=False)
    return pool_index[positions]
```

## 3. Loading the pools

### 3.1 `load_pools` — the points of the sampled pools into memory

Reads every file only once (the sampled pools are grouped by file), keeps only the rows of the sampled pools and turns
every pool into a dict of numpy arrays. Pools smaller than `min_graph_size` are dropped.

```python
def load_pools(sampled_index, use_gse, min_graph_size):
    cols = ["latitude", "longitude", "mean_velocity", "incidence_angle", "seed_pid"]
    gse_cols = [f"gse_{i}" for i in range(64)]
    if use_gse:
        cols += gse_cols
    pools = []
    for (graph_file,), rows in sampled_index.group_by("graph_file"):
        df = pl.read_parquet(graph_file, columns=cols)
        df = df.filter(pl.col("seed_pid").is_in(rows["seed_pid"].implode()))
        pools_in_file = df.partition_by("seed_pid", as_dict=True)
        for pool_df in pools_in_file.values():
            if pool_df.height < min_graph_size:
                continue
            pool = {"lat": pool_df["latitude"].to_numpy(), "lon": pool_df["longitude"].to_numpy(),
                    "vel": pool_df["mean_velocity"].to_numpy().astype(np.float32),
                    "inc": pool_df["incidence_angle"].to_numpy().astype(np.float32),
                    "region": Path(graph_file).parent.parent.name, "orbit": Path(graph_file).parent.name}
            if use_gse:
                pool["gse"] = pool_df.select(gse_cols).to_numpy()
            pools.append(pool)
    return pools
```

## 4. One graph by hand — what `PoolDataset.__getitem__` does

The core of s04: from one pool a **random subset** is drawn, one point of it becomes the **query** (its velocity is
hidden), every point is connected to its **k nearest neighbours**, and the node and edge features are built.

### 4.1 `draw_subset` — which points, which query

```python
def draw_subset(n, subset_min, subset_max, min_graph_size, rng):
    share = rng.uniform(subset_min, subset_max)
    m = min(n, max(min_graph_size, round(share * n)))
    idx = rng.choice(n, size=m, replace=False)
    q = rng.integers(m)
    return idx, q
```

```python
# share = rng.uniform(subset_min, subset_max): a random share of the pool, uniformly between 60 and 90 %
share = rng.uniform(subset_min, subset_max)
print("share =", share)
```

```python
# m = min(n, max(min_graph_size, round(share * n))): the subset size, built from the inside out
print("share * n                         =", share * n)
print("round(share * n)                  =", round(share * n), " <- a whole number of points")
print("max(min_graph_size, ...)          =", max(min_graph_size, round(share * n)), " <- never below the minimum graph size")
m = min(n, max(min_graph_size, round(share * n)))
print("min(n, ...)                       =", m, " <- never more points than the pool has")
```

```python
# idx = rng.choice(n, size=m, replace=False): m different positions out of 0 ... n-1 -> which points are in the graph.
# The order is random, so the first node of the graph is not necessarily the seed
idx = rng.choice(n, size=m, replace=False)
show("idx (positions in the pool)", idx)
```

### 4.2 `pairwise_geodesic` — all distances and bearings

Computes the geodesic distance and bearing between **all pairs** of the subset's points in one `GEOD.inv` call.

```python
def pairwise_geodesic(lat, lon):
    m = len(lat)
    lat1, lon1 = np.repeat(lat, m), np.repeat(lon, m)
    lat2, lon2 = np.tile(lat, m), np.tile(lon, m)
    az12, _, d = GEOD.inv(lon1, lat1, lon2, lat2)
    return d.reshape(m, m), az12.reshape(m, m)
```

**Line by line.** First on 3 points, so every array fits on one line, then on the real subset.

```python
# the inputs: the coordinates of the subset's points (pool["lat"][idx] = the latitudes at the positions idx)
lat, lon = pool["lat"][idx], pool["lon"][idx]
show("lat", lat, precision=6)
show("lon", lon, precision=6)
lat3, lon3 = lat[:3], lon[:3]    # the first 3 points for the small example
m3 = len(lat3)
print("m3 =", m3)
```

```python
# np.repeat(lat, m): every value m times in a row -> the FROM point of every pair
# np.tile(lat, m):   the whole array m times after each other -> the TO point of every pair
# side by side they give every (i, j) combination exactly once: (0,0) (0,1) (0,2) (1,0) (1,1) ...
lat1, lon1 = np.repeat(lat3, m3), np.repeat(lon3, m3)
lat2, lon2 = np.tile(lat3, m3), np.tile(lon3, m3)
print("np.repeat([0, 1, 2], 3):", np.repeat([0, 1, 2], 3))
print("np.tile([0, 1, 2], 3):  ", np.tile([0, 1, 2], 3))
show("lat1 (from)", lat1, precision=6)
show("lat2 (to)", lat2, precision=6)
```

```python
# az12, az21, d = GEOD.inv(lon1, lat1, lon2, lat2): one call for all m*m pairs (pyproj takes lon BEFORE lat).
# az12 = bearing from point 1 to point 2 (degrees, -180 ... 180, 0 = north, clockwise)
# az21 = the back bearing (2 -> 1), not needed: it is az12 of the reversed pair, i.e. the transposed matrix
# d    = distance in metres along the ellipsoid
az12, az21, d = GEOD.inv(lon1, lat1, lon2, lat2)
show("az12", az12, precision=1)
show("d", d, precision=1)
```

```python
# d.reshape(m, m): the flat list of m*m values back into a matrix. Because of repeat/tile the order is row by row:
# row i = from point i, column j = to point j. The diagonal (a point with itself) is 0
show("dist (3 x 3)", d.reshape(m3, m3), precision=1)
show("az (3 x 3)", az12.reshape(m3, m3), precision=1)
```

### 4.3 `knn_edges` — who is connected to whom

Every node is connected to its `k` nearest nodes; then every edge is added in the other direction as well.

```python
def knn_edges(dist, k):
    m = len(dist)
    dist = dist.copy()
    np.fill_diagonal(dist, np.inf)
    neighbours = np.argsort(dist, axis=1)[:, :k]
    target = np.repeat(np.arange(m), k)
    source = neighbours.reshape(-1)
    edge_index = torch.tensor(np.stack([source, target]), dtype=torch.long)
    edge_index = to_undirected(edge_index)
    return edge_index
```

**Line by line** with `k = 3`.

```python
# the inputs
K = 3
k = K
m = len(dist)
print("m =", m)
```

```python
# dist = dist.copy(): work on a copy. fill_diagonal changes the array IN PLACE, and the caller still needs the
# original distances for the edge features
dist_c = dist.copy()
print("same object?", dist_c is dist, "| same values?", np.array_equal(dist_c, dist))
```

```python
# np.fill_diagonal(dist, np.inf): the distance of a node to itself is 0, so every node would be its own nearest
# neighbour. Infinity puts the node itself at the very end of the sorting
np.fill_diagonal(dist_c, np.inf)
show("dist_c, first 4 x 4", dist_c[:4, :4], precision=0)
print("the original is unchanged, diagonal:", np.diag(dist)[:4])
```

```python
# np.argsort(dist, axis=1): per row the column numbers sorted by distance (nearest first). axis=1 = along the row
order = np.argsort(dist_c, axis=1)
show("argsort, first 3 rows", order[:3])
print("row 0: node", order[0, 0], "is nearest to node 0 at", round(dist[0, order[0, 0]]), "m")
```

```python
# [:, :k]: all rows, the first k columns -> the k nearest neighbours of every node
neighbours = order[:, :k]
show("neighbours (row a = the k nodes closest to a)", neighbours[:5])
```

```python
# target = np.repeat(np.arange(m), k): node 0 k times, node 1 k times, ... -> the RECEIVING node of every edge
# source = neighbours.reshape(-1): the neighbour table flattened row by row -> the SENDING node of every edge
target = np.repeat(np.arange(m), k)
source = neighbours.reshape(-1)
show("target", target[:9])
show("source", source[:9])
print("so the first edges are:", [f"{s}->{t}" for s, t in zip(source[:6], target[:6])])
```

### 4.4 `dequantise_gse` — int8 back to the unit-length embedding

AlphaEarth stores every value as int8 (−127 … 127). The values were square-root scaled before rounding, which gives more
resolution near 0; de-quantising squares them again and puts the sign back: `x = (v / 127.5)² · sign(v)`.

```python
def dequantise_gse(v):
    v = v.astype(np.float32)
    x = (v / 127.5) ** 2 * np.sign(v)
    return x
```

```python
# the formula piece by piece
print("v / 127.5           :", (v / 127.5)[0, :6], " <- scaled to -1 ... 1")
print("(v / 127.5) ** 2    :", ((v / 127.5) ** 2)[0, :6], " <- squared, the sign is lost")
print("np.sign(v)          :", np.sign(v)[0, :6], " <- -1, 0 or +1")
x_gse = (v / 127.5) ** 2 * np.sign(v)
print("(...) ** 2 * sign(v):", x_gse[0, :6], " <- the sign back")
print("\nvector length per point (should be ~1):", np.linalg.norm(x_gse, axis=1)[:5])
print("check, same as the function:", np.array_equal(x_gse, dequantise_gse(pool["gse"][idx])))
```

### 4.5 `build_node_features` — the node matrix `x`

One row per node: `[normalized velocity (0 at the query), is_query, normalized incidence]`, plus the 64 GSE values.

```python
def build_node_features(pool, idx, q, stats, use_gse):
    m = len(idx)
    v_stats, i_stats = stats["mean_velocity"], stats["incidence_angle"]
    vel = (pool["vel"][idx] - v_stats["mean"]) / v_stats["std"]
    vel[q] = 0.0
    is_query = np.zeros(m, dtype=np.float32)
    is_query[q] = 1.0
    inc = (pool["inc"][idx] - i_stats["mean"]) / i_stats["std"]
    gse = dequantise_gse(pool["gse"][idx]) if use_gse else None
    cols = [vel, is_query, inc]
    x = np.column_stack(cols + ([gse] if use_gse else []))
    return torch.tensor(x, dtype=torch.float32)
```

### 4.6 `build_edge_features` — one row per edge

`[log distance (normalized once the edge stats exist), sin(bearing), cos(bearing)]`, the bearing taken from the
receiving to the sending node.

```python
def build_edge_features(dist, az, edge_index, stats):
    src, dst = edge_index.numpy()
    d = np.log1p(dist[src, dst])
    if "edge_dist" in stats:
        d = (d - stats["edge_dist"]["mean"]) / stats["edge_dist"]["std"]
    bearing = np.radians(az[dst, src])
    sin, cos = np.sin(bearing), np.cos(bearing)
    edge_attr = np.column_stack([d, sin, cos])
    return torch.tensor(edge_attr, dtype=torch.float32)
```

## 5. Packed into a class

### 5.1 `PoolDataset` — section 4 in `__getitem__`

`PoolDataset` stores the pools and settings; `__getitem__(i)` runs exactly the steps of section 4 for pool `i`. The
random generator decides the subset and query:
- **train**: seeded with `(seed, epoch, i)` → a new draw every epoch, but reproducible, and independent of which
  DataLoader worker builds the graph;
- **dev** (`fixed_draws=True`): seeded with `i` → pool `i` always gives the same graph, so the dev loss of different
  epochs is comparable.

```python
class PoolDataset(torch.utils.data.Dataset):
    def __init__(self, pools, stats, k_graph, subset_min, subset_max, min_graph_size, use_gse, fixed_draws=False, seed=0):
        self.seed = seed
        self.epoch = 0
        self.pools = pools
        self.stats = stats
        self.k_graph = k_graph
        self.subset_min = subset_min
        self.subset_max = subset_max
        self.min_graph_size = min_graph_size
        self.use_gse = use_gse
        self.fixed_draws = fixed_draws

    def set_epoch(self, epoch):
        self.epoch = epoch

    def __len__(self):
        return len(self.pools)

    def __getitem__(self, i):
        pool = self.pools[i]
        n = len(pool["vel"])
        lat, lon = pool["lat"], pool["lon"]
        rng = np.random.default_rng(i) if self.fixed_draws else np.random.default_rng([self.seed, self.epoch, i])
        idx, q = draw_subset(n, self.subset_min, self.subset_max, self.min_graph_size, rng)
        dist, az = pairwise_geodesic(lat[idx], lon[idx])
        edge_index = knn_edges(dist, self.k_graph)
        x = build_node_features(pool, idx, q, self.stats, self.use_gse)
        edge_attr = build_edge_features(dist, az, edge_index, self.stats)
        y = torch.tensor([(pool["vel"][idx][q] - self.stats["mean_velocity"]["mean"]) / self.stats["mean_velocity"]["std"]], dtype=torch.float32)
        query_mask = torch.zeros(len(idx), dtype=torch.bool)
        query_mask[q] = True
        return Data(x=x, edge_index=edge_index, edge_attr=edge_attr, y=y, query_mask=query_mask,
            region=pool["region"], orbit=pool["orbit"])
```

## 7. The model — `InSARGNN`

Follow the shapes: `(nodes, features)` → `(nodes, hidden)` through the encoder and the message passing layers →
`(graphs, hidden)` at the query nodes → one `mu` and one `logvar` per graph.

```python
class InSARGNN(nn.Module):
    def __init__(self, n_node_features, edge_dim, hidden_dim, n_layers, heads, dropout):
        super().__init__()
        self.encoder = nn.Sequential(nn.Linear(n_node_features, hidden_dim), nn.GELU(), nn.Linear(hidden_dim, hidden_dim))
        self.convs = nn.ModuleList([
            GATv2Conv(hidden_dim, hidden_dim // heads, heads=heads, edge_dim=edge_dim, dropout=dropout)
            for _ in range(n_layers)])
        self.norms = nn.ModuleList([nn.LayerNorm(hidden_dim) for _ in range(n_layers)])
        self.head_mu = nn.Linear(hidden_dim, 1)
        self.head_logvar = nn.Linear(hidden_dim, 1)

    def forward(self, data):
        h = self.encoder(data.x)
        for conv, norm in zip(self.convs, self.norms):
            h = norm(h + nn.functional.gelu(conv(h, data.edge_index, data.edge_attr)))
        h_query = h[data.query_mask]
        mu = self.head_mu(h_query).squeeze(-1)
        logvar = self.head_logvar(h_query).squeeze(-1).clamp(-8, 8)
        return mu, logvar
```

**`__init__` line by line.** `super().__init__()` sets up `nn.Module`; every layer assigned to `self` afterwards is registered, so its weights are trained, moved to the GPU and saved.

## 8. The loss — `loss_fn`

Heteroscedastic Gaussian NLL: `0.5 · (logvar + (y − mu)² / exp(logvar))`. A large error is cheaper if the model also says
"I'm unsure" (large `logvar`), but being unsure costs `logvar` itself.

```python
def loss_fn(mu, logvar, y):
    loss = 0.5 * (logvar + (y - mu) ** 2 / torch.exp(logvar))
    return loss.mean()
```

```python
# three toy queries: right and confident, wrong and confident, wrong but unsure
y_ = torch.tensor([1.0, 1.0, 1.0])
mu_ = torch.tensor([1.0, 0.0, 0.0])
logvar_ = torch.tensor([-3.0, -3.0, 1.0])
print("y - mu            :", (y_ - mu_))
print("(y - mu) ** 2     :", (y_ - mu_) ** 2, " <- squared error")
print("exp(logvar)       :", torch.exp(logvar_), " <- the predicted variance sigma^2")
print("error / variance  :", (y_ - mu_) ** 2 / torch.exp(logvar_), " <- the error measured in units of the predicted variance")
per_query = 0.5 * (logvar_ + (y_ - mu_) ** 2 / torch.exp(logvar_))
print("0.5 * (logvar + .):", per_query, " <- logvar is the price for being unsure")
print(".mean()           :", per_query.mean(), "| check:", torch.isclose(per_query.mean(), loss_fn(mu_, logvar_, y_)).item())
```

```python
y1 = torch.tensor([1.0])
for m_, lv in [(1.0, 0.0), (1.0, -3.0), (0.0, 0.0), (0.0, -3.0), (0.0, 1.0), (0.0, 3.0)]:
    print(f"error {abs(1 - m_):.0f}, logvar {lv:4.1f} (sigma {np.exp(lv / 2):5.2f}) -> loss {loss_fn(torch.tensor([m_]), torch.tensor([lv]), y1).item():6.3f}")
print("-> right and confident is best; wrong and confident is worst; wrong but unsure is in between")
```

### 9.2 `train_one_epoch` — one pass over the train loader

```python
def train_one_epoch(model, loader, optimizer, grad_clip, device):
    model.train()
    total_loss, n_graphs = 0.0, 0
    for batch in loader:
        batch = batch.to(device)
        optimizer.zero_grad()
        mu, logvar = model(batch)
        loss = loss_fn(mu, logvar, batch.y)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), grad_clip)
        optimizer.step()
        total_loss += loss.item() * batch.num_graphs
        n_graphs += batch.num_graphs
    return total_loss / n_graphs
```

**Line by line** for one batch.

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
set_seed(42)
model = InSARGNN(n_node_features, edge_dim, hidden_dim=32, n_layers=2, heads=4, dropout=0.1).to(device)
train_loader = DataLoader(train_ds, batch_size=4, shuffle=True)
dev_loader = DataLoader(dev_ds, batch_size=4, shuffle=False)
# AdamW: the optimizer, it gets the parameters it is allowed to change, the learning rate and the weight decay
optimizer = torch.optim.AdamW(model.parameters(), lr=3e-3, weight_decay=1e-4)
grad_clip = 5.0
```

```python
# model.train(): training mode -> dropout is active (model.eval() switches it off)
model.train()
print("model.training =", model.training)
total_loss, n_graphs = 0.0, 0
```

```python
# for batch in loader: batch = batch.to(device) -> all tensors of the batch to the GPU (here: wherever device is)
batch = next(iter(train_loader)).to(device)
print(batch)
```

```python
# optimizer.zero_grad(): deletes the gradients of the previous batch. PyTorch ADDS new gradients to existing ones,
# so without this line every step would also contain all earlier batches
optimizer.zero_grad()
w = model.head_mu.weight
print("gradient of head_mu.weight before backward:", w.grad)
```

```python
# forward pass and loss: what the model predicts for the queries and how wrong (and how honest) that is
mu, logvar = model(batch)
loss = loss_fn(mu, logvar, batch.y)
print("mu:", mu.detach().numpy(), "\ny: ", batch.y.numpy(), "\nloss:", loss.item())
```

```python
# loss.backward(): backpropagation, computes d(loss)/d(weight) for every weight and stores it in weight.grad
loss.backward()
print("gradient of head_mu.weight:", tuple(w.grad.shape), "| norm:", w.grad.norm().item())
```

```python
# clip_grad_norm_(parameters, max_norm): if the length of ALL gradients together (one long vector) is above
# grad_clip, they are scaled down to exactly that length. Protects against single huge steps. Returns the length before
total_norm = torch.nn.utils.clip_grad_norm_(model.parameters(), grad_clip)
print("total gradient norm before clipping:", total_norm.item(), "| clipped:", total_norm.item() > grad_clip)
```

```python
# loss.item(): the tensor as a plain python float. loss is the MEAN over the batch, so it is multiplied by the batch
# size before summing; at the end total / n_graphs = mean per graph, and a smaller last batch counts less
total_loss += loss.item() * batch.num_graphs
n_graphs += batch.num_graphs
print("total_loss =", total_loss, "| n_graphs =", n_graphs, "| mean so far =", total_loss / n_graphs)
```

### 9.3 `evaluate` — predictions and metrics without training

```python
def evaluate(model, loader, device):
    model.eval()
    with torch.no_grad():
        mus, logvars, ys, regions, orbits = [], [], [], [], []
        for batch in loader:
            batch = batch.to(device)
            mu, logvar = model(batch)
            mus.append(mu.cpu()); logvars.append(logvar.cpu()); ys.append(batch.y.cpu())
            regions.extend(batch.region); orbits.extend(batch.orbit)
        mu, logvar, y = torch.cat(mus), torch.cat(logvars), torch.cat(ys)
        nll = loss_fn(mu, logvar, y).item()
        v = loader.dataset.stats["mean_velocity"]
        mu_mm = mu * v["std"] + v["mean"]
        y_mm = y * v["std"] + v["mean"]
        sigma_mm = torch.exp(0.5 * logvar) * v["std"]
        err = mu_mm - y_mm
        rmse = torch.sqrt((err ** 2).mean()).item()
        mae = err.abs().mean().item()
        inside = (err.abs() <= 1.645 * sigma_mm).float().mean().item()
        metrics = {"nll": nll, "rmse": rmse, "mae": mae, "inside_90": inside}
        predictions = pl.DataFrame({"region": regions, "orbit": orbits, "y_mm": y_mm.numpy(),
                            "mu_mm": mu_mm.numpy(), "sigma_mm": sigma_mm.numpy()})
    return metrics, predictions
```

```python
# model.eval(): dropout off. torch.no_grad(): no gradients are recorded (faster, less memory, nothing to train here)
model.eval()
with torch.no_grad():
    out_mu, _ = model(next(iter(dev_loader)).to(device))
print("model.training =", model.training, "| output records gradients:", out_mu.requires_grad)
```

```python
# the loop collects the outputs of every batch in lists. .cpu(): back from the GPU. extend (not append) adds the
# single strings of batch.region instead of the whole list as one element
mus, logvars, ys, regions, orbits = [], [], [], [], []
with torch.no_grad():
    for b in dev_loader:
        b = b.to(device)
        m_, lv_ = model(b)
        mus.append(m_.cpu()); logvars.append(lv_.cpu()); ys.append(b.y.cpu())
        regions.extend(b.region); orbits.extend(b.orbit)
print("batches:", len(mus), "| shapes per batch:", [tuple(t.shape) for t in mus], "| regions:", len(regions))
```

```python
# torch.cat: the per-batch tensors into one tensor -> one entry per query (not per batch)
mu, logvar, y = torch.cat(mus), torch.cat(logvars), torch.cat(ys)
print("mu:", tuple(mu.shape), "| nll (normalized units):", loss_fn(mu, logvar, y).item())
```

```python
# back to mm/yr: undo the normalization. sigma = exp(0.5 * logvar) = sqrt(variance), scaled with the std as well
v = dev_loader.dataset.stats["mean_velocity"]
mu_mm = mu * v["std"] + v["mean"]
y_mm = y * v["std"] + v["mean"]
sigma_mm = torch.exp(0.5 * logvar) * v["std"]
print("mu_mm:", mu_mm.numpy()[:4], "\ny_mm: ", y_mm.numpy()[:4], "\nsigma:", sigma_mm.numpy()[:4])
```

```python
# the metrics: RMSE punishes large errors more (squared), MAE is the typical error.
# inside_90: share of queries whose true value lies within mu +- 1.645 sigma (the 90 % interval of a Gaussian)
err = mu_mm - y_mm
print("err:", err.numpy()[:4])
print("rmse:", torch.sqrt((err ** 2).mean()).item(), "| mae:", err.abs().mean().item())
inside_mask = err.abs() <= 1.645 * sigma_mm
print("inside the interval:", inside_mask.tolist()[:8], "-> .float().mean() =", inside_mask.float().mean().item())
```

```python
# the predictions table: one row per query, used for the per-region and per-orbit metrics
metrics, dev_predictions = evaluate(model, dev_loader, device)
print({k: round(v_, 3) for k, v_ in metrics.items()})
dev_predictions.head()
```

### 9.4 `save_checkpoint` and `train_model` — the epoch loop with early stopping

Every epoch: new train draws (`set_epoch`), one training pass, one evaluation on dev. A new best dev NLL saves a
checkpoint; after `patience` epochs without a new best the training stops.

```python
def save_checkpoint(model, params, epoch, path):
    path.parent.mkdir(parents=True, exist_ok=True)
    torch.save({"state_dict": model.state_dict(), "params": params, "epoch": epoch}, path)

def train_model(model, train_loader, dev_loader, lr, weight_decay, max_epochs, patience, grad_clip,
                device, checkpoint_path, params):
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)
    best_loss, best_epoch, epochs_without_improvement, history = float("inf"), None, 0, []
    for epoch in range(max_epochs):
        train_loader.dataset.set_epoch(epoch)
        train_loss = train_one_epoch(model, train_loader, optimizer, grad_clip, device)
        dev_metrics, _ = evaluate(model, dev_loader, device)
        history.append({"epoch": epoch, "train_loss": train_loss,
                "dev_loss": dev_metrics["nll"], "dev_rmse": dev_metrics["rmse"]})
        print(history[-1], flush=True)
        if dev_metrics["nll"] < best_loss:
            best_loss, best_epoch = dev_metrics["nll"], epoch
            epochs_without_improvement = 0
            save_checkpoint(model, params, epoch, checkpoint_path)
        else:
            epochs_without_improvement += 1
            if epochs_without_improvement >= patience:
                break
    return history, best_epoch
```

```python
# save_checkpoint: model.state_dict() = all weights as an ordered dict {layer name: tensor}. Saved together with the
# params (to rebuild the model later) and the epoch. mkdir(parents=True, exist_ok=True): create missing folders, no
# error if they exist
sd = model.state_dict()
print("state_dict:", type(sd).__name__, "with", len(sd), "tensors, e.g.", list(sd)[:3])
```

```python
# the bookkeeping of the early stopping:
# best_loss = inf so that the first epoch is always an improvement; best_epoch = None until then
best_loss, best_epoch, epochs_without_improvement, history = float("inf"), None, 0, []
print("float('inf') > any number:", float("inf") > 1e30)
```

```python
# the whole training on the toy data (small patience so you can see it stop)
checkpoint_path = Path(tempfile.mkdtemp()) / "checkpoints" / "toy_run" / "best_model.pt"
set_seed(42)
model = InSARGNN(n_node_features, edge_dim, hidden_dim=32, n_layers=2, heads=4, dropout=0.1).to(device)
history, best_epoch = train_model(model, train_loader, dev_loader, lr=3e-3, weight_decay=1e-4, max_epochs=60,
                                  patience=8, grad_clip=5.0, device=device, checkpoint_path=checkpoint_path,
                                  params={"stats": stats})
print("epochs run:", len(history), "| best epoch:", best_epoch)
```

### 9.5 `evaluate_on_dev` — reload the best epoch and compute the metrics per group

```python
# torch.load: the saved dict back. map_location=device: load the tensors directly onto this device
checkpoint = torch.load(checkpoint_path, map_location=device)
print("checkpoint keys:", list(checkpoint), "| saved at epoch", checkpoint["epoch"])
# load_state_dict: overwrites the current (last-epoch) weights with the best ones
model.load_state_dict(checkpoint["state_dict"])
train_metrics, _ = evaluate(model, train_loader, device)
dev_metrics, dev_predictions = evaluate(model, dev_loader, device)
print("train:", {k: round(v_, 3) for k, v_ in train_metrics.items()})
print("dev:  ", {k: round(v_, 3) for k, v_ in dev_metrics.items()})
```

```python
# per_group(col): the error as a polars expression (computed per row once it is used), then per group:
# pl.len() = number of rows, (err ** 2).mean().sqrt() = RMSE, err.abs().mean() = MAE, the inside share as a mean of booleans
err = pl.col("mu_mm") - pl.col("y_mm")
per_region = (dev_predictions.group_by("region")
              .agg(pl.len().alias("n"),
                   (err ** 2).mean().sqrt().alias("rmse"),
                   err.abs().mean().alias("mae"),
                   (err.abs() <= 1.645 * pl.col("sigma_mm")).mean().alias("inside_90"))
              .sort("region"))
print(per_region)
# .to_dicts(): one dict per row, the format the markdown tables of the log are built from
print(per_region.to_dicts())
```
